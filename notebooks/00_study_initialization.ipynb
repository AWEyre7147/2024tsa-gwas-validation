### Study Purpose and Design

This study asks whether genemoic regions associate with success vs. elimination status in a 2024TSA Labrador Retriever + German Shorthaired Pointer explosive detection training dog cohort. In addition to searching for associations within each breed, the study examines whether combining breeds changes the interpretation. The analyses provide supporting, exploratory evidence. An association identifies regions worth studying, but do not by themselves establish that a variant casues a behavioral outcome.

Additionally, the study will use the 2024 Labrator Retrivers to replicate + validate the genetic regions identified in the [2013TSA significant-hit reference](../docs/references/2013TSA%20Significant%20Hits.md).

### Define Scheme A and Scheme B

The outcome of interest is elimination from the training program for behavioral reasons. These reasons are treated as one category rather than divided into subtypes.

**Scheme A**, uses dogs recorded as Validated or Certified, with no elimination reason, as controls. These dogs are compared with dogs eliminated for behavioral reasons as cases.

**Scheme B**, adds dogs eliminated for Medical, Environmental, or Inconsistency reasons to the control group. The dogs eliminated for behavioral reasons stay the same. This second comparison asks whether the findings change when the definition of the control group is broadened. It does not assume that the additional elimination reasons are unrelated to behavior.

| Scheme | Cases | Controls |
| :---: | :---: | :---: |
| A | Behavioral Elimination | Validated, Certified |
| B | Behavioral Elimination | Validated, Certified, Medical Elimination, Environmental Elimination, Inconsistency |
