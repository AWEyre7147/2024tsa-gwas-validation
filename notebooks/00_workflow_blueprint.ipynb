# Study purpose and design

This study asks whether genetic regions associated with training outcomes in the earlier 2013TSA Labrador Retriever study show evidence of association in an independent group of dogs, the 2024TSA cohort. Testing a previous finding in a separate cohort is called replication. It is the main purpose of this study.

The replication analysis will use Labrador Retrievers and the genetic regions identified in the [2013TSA significant-hit reference](../docs/references/2013TSA%20Significant%20Hits.md). We will define what counts as evidence of replication before examining the association results.

The cohort also includes German Shorthaired Pointers. We plan to search for additional associations within each breed and examine whether combining breeds changes the interpretation. These analyses will provide supporting, exploratory evidence. An association can identify a region worth studying, but does not by itself establish that a variant causes a behavioral outcome.


## Comparing training outcomes

The outcome of interest is elimination from training for a recorded behavioral reason. Dogs with this outcome form the case group. Behavioral elimination reasons are treated as one category rather than divided into subtypes.

The main comparison, called **Scheme A**, uses dogs recorded as Validated or Certified, with no elimination reason, as controls. These dogs are compared with the behavioral cases.

We will also make a broader comparison, **Scheme B**, that adds dogs eliminated for Medical, Environmental, or Inconsistency reasons to the control group. The behavioral cases stay the same. This second comparison asks whether the findings change when the definition of the control group is broadened. It does not assume that every other elimination is unrelated to behavior.

Scheme A is the primary replication analysis. Scheme B is a supporting check on how the choice of controls affects the findings. We will retain these roles regardless of which comparison produces stronger associations.

Breed, sex, and training outcomes come from the January 2024 records. We use these records on the investigator's assumption that earlier sample-record corrections were completed. Dogs with unresolved outcomes or uncertain links between their records and genetic samples are excluded. [Notebook 01](01_inputs_and_phenotypes.ipynb) describes the study groups and their sizes.


## How the study will proceed

After preparing the study groups, we will examine the quality of the genetic measurements and assess whether the available dogs can support the planned analyses. The small number of behavioral cases will be an important limitation when estimating associations and interpreting uncertainty.

We also intend to evaluate whether reference genetic data can help estimate variants that were not measured directly. These estimated variants will be used only if their accuracy is supported by validation. The replication analysis can still use suitable directly measured variants if this additional step is unsuccessful.

We will then test the earlier Labrador findings, followed by the exploratory analyses within each breed and the combined-breed comparison. Selected regions may be examined for nearby genes and relevant biological evidence. This follow-up will help prioritize candidates for further research.

The study groups have been prepared and autosomal marker filtering is complete. Genetic structure and sharing among dogs are described in Notebook 03; the association model remains to be selected. Later chapters will explain each method and its findings as the work progresses.
